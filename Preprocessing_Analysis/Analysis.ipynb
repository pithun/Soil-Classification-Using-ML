```python
import os
import pandas as pd
```

```python
all_csvs = os.listdir('.')
all_csvs=[a for a in all_csvs if 'csv' in a.lower()]
```

```python
all_csvs
```

```python
def csv_concater(list_of_csvs):
    df1=pd.read_csv(list_of_csvs[0])
    for csv in list_of_csvs[1:]:
        df2=pd.read_csv(csv)
        df1=pd.concat([df1, df2], axis=0)
        #print(df1.shape)
    return df1
        
```

```python
metrics=csv_concater(all_csvs)
```

```python
metrics.sort_values('accuracy_test', ascending=False)['params'].values

#best performing set of params (10, (0.1, 0.9), (100, 300), 2.5, -0.5, 50, 4, 45, 30)
```
